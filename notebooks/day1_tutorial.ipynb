# 第一天实操教程：MOF 结构可视化与特征提取

本教程涵盖：
1. MOF 结构可视化
2. 格式转换
3. 几何特征提取
4. 性能参数计算

## 1. 环境准备

```python
import sys
sys.path.append('../src')

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path

# 导入我们的工具
from visualization.mof_visualizer import MOFVisualizer
from feature_extraction.geometric_features import SimplifiedGeometricExtractor
from calculations.property_calculator import PropertyCalculator

print("环境准备完成！")
```

## 2. MOF 结构可视化

```python
# 加载示例 MOF 结构
example_cif = '../data/examples/example_mof.cif'

# 创建可视化器
viz = MOFVisualizer(example_cif)

# 打印结构信息
viz.print_info()
```

```python
# 3D 可视化 (在 Jupyter Notebook 中显示)
view = viz.view_3d(style='sphere', show_cell=True)
view.show()
```

```python
# 2D 投影图
fig, ax = viz.plot_2d_projection(axis='z', repeat=(1, 1, 1))
plt.show()
```

## 3. 格式转换

```python
# 导出多种格式
output_dir = Path('../data/examples/converted')
output_dir.mkdir(exist_ok=True, parents=True)

exported = viz.export_formats(str(output_dir / 'example_mof'))

for fmt, filepath in exported.items():
    print(f"{fmt}: {filepath}")
```

## 4. 几何特征提取

```python
# 使用简化提取器 (不需要 Zeo++)
extractor = SimplifiedGeometricExtractor()
features = extractor.calculate_features(example_cif)

# 显示特征
print("几何特征:")
for key, value in features.items():
    print(f"  {key}: {value:.3f}" if isinstance(value, float) else f"  {key}: {value}")
```

## 5. 性能参数计算

```python
# 创建性能计算器
calc = PropertyCalculator()

# 计算所有性能 (这可能需要几分钟)
properties = calc.calculate_all_properties(example_cif, n_samples=2000)

# 打印结果
calc.print_properties(properties)
```

## 6. 孔径分布可视化

```python
# 绘制孔径分布
if 'psd_radii' in properties and 'psd_distribution' in properties:
    fig, ax = calc.psd_calc.plot_psd(properties['psd_radii'], properties['psd_distribution'])
    plt.show()
else:
    print("孔径分布数据不可用")
```

## 7. 批量处理示例

```python
# 如果有多个 CIF 文件，可以批量处理
from pathlib import Path
from feature_extraction.geometric_features import batch_extract_features

# 获取所有 CIF 文件
cif_dir = Path('../data/examples')
cif_files = list(cif_dir.glob('*.cif'))

print(f"发现 {len(cif_files)} 个 CIF 文件")

# 批量提取特征
if len(cif_files) > 0:
    df = batch_extract_features(cif_files, use_zeopp=False)
    print("\n特征提取完成！")
    display(df.head())
```

## 8. 数据保存

```python
# 保存特征数据
if len(cif_files) > 0:
    output_csv = '../data/examples/mof_features.csv'
    df.to_csv(output_csv, index=False)
    print(f"特征数据已保存: {output_csv}")
```
